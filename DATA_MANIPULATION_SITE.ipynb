```python
from pyspark.sql import functions as f

df = spark.read.table("workspace.default.row_data_site")

df_out = df.withColumn( "Minuti_tot_sul_Sito", f.when( 
                    f.col("Durata_Visita_Sito") <= 60, f.round(f.col("Durata_Visita_Sito") / 100, 2) 
                      ).otherwise(f.round(f.col("Durata_Visita_Sito") / 60, 2) ) )

display(df_out)

(df_out.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .partitionBy("Data_Invio_Statistiche")
    .saveAsTable("default.bronze_data_site"))
```
